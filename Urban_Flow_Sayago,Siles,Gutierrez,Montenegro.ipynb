````python
# librerías que voy a usar
import pandas as pd
import numpy as np
````

````python
# cargo el dataset desde el link
url = "https://raw.githubusercontent.com/HAD141/datasets/refs/heads/main/TrabajosPracticos/urban_flow/speeding_fines.csv"
df = pd.read_csv(url)

# veo las primeras filas para entender qué tiene
df.head()
````

````python
# paso las fechas a formato datetime (si hay errores los convierte en NaT)
df["fecha"] = pd.to_datetime(df["fecha"], errors="coerce")

# cambio las fechas inválidas por una fija
df["fecha"] = df["fecha"].fillna(pd.Timestamp("1932-01-01"))

# dejo el formato como pide el ejercicio
df["fecha"] = df["fecha"].dt.strftime("%Y-%m-%d")

df["fecha"].head(10)
````

````python
# intento convertir las horas
df["hora"] = pd.to_datetime(df["hora"], errors="coerce")

# cambio al formato 24hs
df["hora"] = df["hora"].dt.strftime("%H:%M")

# si hay errores, pongo 00:00
df["hora"] = df["hora"].fillna("00:00")

df["hora"].head(10)
````

````python
# limpio caracteres raros y paso todo a mayúsculas
df["ubicacion"] = df["ubicacion"].str.replace(r"[^a-zA-Z0-9 ]", "", regex=True)
df["ubicacion"] = df["ubicacion"].str.upper()

df["ubicacion"].head(10)
````

````python
# saco símbolos y dejo solo letras/números
df["patente"] = df["patente"].str.replace(r"[^a-zA-Z0-9]", "", regex=True)

# paso a mayúsculas
df["patente"] = df["patente"].str.upper()

# si queda vacío lo marco como NA
df["patente"] = df["patente"].replace("", pd.NA)

df["patente"].tail(10)
````

````python
# columnas que veo importantes para la multa
cols_importantes = [
    "fecha",
    "hora",
    "ubicacion",
    "patente",
    "velocidad_registrada",
    "velocidad_maxima"
]

antes = len(df)

# elimino filas con datos faltantes en esas columnas
df = df.dropna(subset=cols_importantes)

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
print("Las columnas observadas son:")
for c in cols_importantes:
    print("-", c)
````

````python
# elimino valores raros (velocidades poco realistas)
antes = len(df)

df = df[
    (df["velocidad_registrada"] < 200) &
    (df["velocidad_maxima"] < 130)
]

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
print("Las columnas observadas son:")
print("- velocidad_registrada")
print("- velocidad_maxima")
````

````python
# diferencia directa entre velocidad registrada y máxima
df["exceso_velocidad_real"] = df["velocidad_registrada"] - df["velocidad_maxima"]

df[["patente", "exceso_velocidad_real"]].head(10)
````

````python
# le sumo un 5% de tolerancia a la velocidad máxima
df["exceso_velocidad"] = df["velocidad_registrada"] - (df["velocidad_maxima"] * 1.05)

df[["patente", "exceso_velocidad"]].head(10)
````

````python
# dejo solo los que realmente superan el límite
antes = len(df)

df = df[df["exceso_velocidad"] > 0]

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
````

````python
# guardo el dataset limpio
import os

os.makedirs("urban_flow/data/interim", exist_ok=True)

df.to_csv("urban_flow/data/interim/speeding_fines.csv", index=False)
````
