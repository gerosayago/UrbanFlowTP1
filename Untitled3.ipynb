```python
# librerías que voy a usar
import pandas as pd
import numpy as np
```

```python
# cargo el dataset desde el link
url = "https://raw.githubusercontent.com/HAD141/datasets/refs/heads/main/TrabajosPracticos/urban_flow/speeding_fines.csv"
df = pd.read_csv(url)

# veo las primeras filas para entender qué tiene
df.head()
```

```python
# paso las fechas a formato datetime (si hay errores los convierte en NaT)
df["fecha"] = pd.to_datetime(df["fecha"], errors="coerce")

# reemplazo las fechas inválidas por una fija
df["fecha"] = df["fecha"].fillna(pd.Timestamp("1932-01-01"))

# dejo el formato como pide el ejercicio
df["fecha"] = df["fecha"].dt.strftime("%Y-%m-%d")

df["fecha"].head(10)
```

```python
# intento convertir las horas
df["hora"] = pd.to_datetime(df["hora"], errors="coerce")

# paso al formato 24hs
df["hora"] = df["hora"].dt.strftime("%H:%M")

# si hubo errores, pongo 00:00
df["hora"] = df["hora"].fillna("00:00")

df["hora"].head(10)
```

```python
# limpio caracteres raros y paso todo a mayúsculas
df["ubicacion"] = df["ubicacion"].str.replace(r"[^a-zA-Z0-9 ]", "", regex=True)
df["ubicacion"] = df["ubicacion"].str.upper()

df["ubicacion"].head(10)
```

```python
# saco símbolos y dejo solo letras/números
df["patente"] = df["patente"].str.replace(r"[^a-zA-Z0-9]", "", regex=True)

# paso a mayúsculas
df["patente"] = df["patente"].str.upper()

# si queda vacío lo marco como NA
df["patente"] = df["patente"].replace("", pd.NA)

df["patente"].tail(10)
```

```python
# columnas que considero importantes para la multa
cols_importantes = [
    "fecha",
    "hora",
    "ubicacion",
    "patente",
    "velocidad_registrada",
    "velocidad_maxima"
]

antes = len(df)

# elimino filas con datos faltantes en esas columnas
df = df.dropna(subset=cols_importantes)

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
print("Las columnas observadas son:")
for c in cols_importantes:
    print("-", c)
```

```python
# elimino valores raros (velocidades poco realistas)
antes = len(df)

df = df[
    (df["velocidad_registrada"] < 200) &
    (df["velocidad_maxima"] < 130)
]

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
print("Las columnas observadas son:")
print("- velocidad_registrada")
print("- velocidad_maxima")
```

```python
# diferencia directa entre velocidad registrada y máxima
df["exceso_velocidad_real"] = df["velocidad_registrada"] - df["velocidad_maxima"]

df[["patente", "exceso_velocidad_real"]].head(10)
```

```python
# le sumo un 5% de tolerancia a la velocidad máxima
df["exceso_velocidad"] = df["velocidad_registrada"] - (df["velocidad_maxima"] * 1.05)

df[["patente", "exceso_velocidad"]].head(10)
```

```python
# dejo solo los que realmente superan el límite
antes = len(df)

df = df[df["exceso_velocidad"] > 0]

despues = len(df)

print("Se eliminaron", antes - despues, "filas.")
```

```python
# guardo el dataset limpio
import os

os.makedirs("urban_flow/data/interim", exist_ok=True)

df.to_csv("urban_flow/data/interim/speeding_fines.csv", index=False)
```

Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

Reciba el DataFrame limpio como argumento de inicialización.

Encapsule correctamente los datos.

Implemente métodos para:

Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.

Exceso promedio de velocidad. Debe retorna un flotante.

Exceso real promedio de velocidad . Debe retorna un flotante.

Contabilizar la cantidad de multas por cada ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

Crear el objecto e invocar cada uno de los métodos en celdas separadas.

```python
## Ejercicio 04 - Análisis de multas con clase
### Definición de la clase FineAnalyzer
##importo libreria pandas
import pandas as pd

class FineAnalyzer:
    def __init__(self, df: pd.DataFrame):
        # guardo el dataframe limpio
        self.df = df.copy()

    def top_patentes(self) -> pd.DataFrame:
        # cuento cuántas veces aparece cada patente
        ranking = self.df["patente"].value_counts().head(5)

        # paso a dataframe
        ranking = ranking.reset_index()
        ranking.columns = ["patente", "cantidad"]

        # index desde 1
        ranking.index = range(1, len(ranking) + 1)

        return ranking

    def top_horas(self) -> pd.DataFrame:
        # cuento multas por hora
        ranking = self.df["hora"].value_counts().head(5)

        ranking = ranking.reset_index()
        ranking.columns = ["hora", "cantidad"]
        ranking.index = range(1, len(ranking) + 1)

        return ranking

    def exceso_promedio(self) -> float:
        # promedio del exceso con margen
        return self.df["exceso_velocidad"].mean()

    def exceso_real_promedio(self) -> float:
        # promedio del exceso real
        return self.df["exceso_velocidad_real"].mean()

    def multas_por_ubicacion(self) -> pd.DataFrame:
        # agrupo por ubicación
        resultado = self.df["ubicacion"].value_counts()

        resultado = resultado.reset_index()
        resultado.columns = ["ubicacion", "cantidad"]

        return resultado
```

```python
### Uso de la clase
# creo el objeto usando el dataframe limpio del ejercicio 3
analizador = FineAnalyzer(df)
### Top 5 patentes más multadas
analizador.top_patentes()
### Top 5 horarios con más multas
analizador.top_horas()
### Exceso promedio de velocidad
analizador.exceso_promedio()
### Exceso real promedio
analizador.exceso_real_promedio()
### Multas por ubicación
analizador.multas_por_ubicacion()
```

```python
### Importación de librerías para gráficos
import matplotlib.pyplot as plt
```

### Top 10 patentes más reincidentes

```python
top10 = df["patente"].value_counts().head(10)

top10.plot(kind="bar")
plt.title("Top 10 patentes más multadas")
plt.xlabel("Patente")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)

plt.savefig("urban_flow/data/interim/plots/fines.jpg")
plt.show()
```

### Porcentaje de infracciones por hora

```python
horas = df["hora"].value_counts()

horas.plot(kind="pie", autopct="%1.1f%%")
plt.title("Porcentaje de multas por hora")
plt.ylabel("")

plt.savefig("urban_flow/data/interim/plots/hours.jpg")
plt.show()
```

### Multas por mes

```python
# extraigo mes
df["mes"] = pd.to_datetime(df["fecha"]).dt.month

meses = df["mes"].value_counts().sort_values(ascending=True)

meses.plot(kind="barh")
plt.title("Multas por mes")
plt.xlabel("Cantidad")

plt.savefig("urban_flow/data/interim/plots/months.jpg")
plt.show()
```

```python
### Exceso de velocidad por hora 00:00
filtro_hora = df[df["hora"] == "00:00"]

filtro_hora.groupby("hora")["exceso_velocidad"].mean().plot()

plt.title("Exceso promedio a las 00:00")

plt.savefig("urban_flow/data/interim/plots/hour.jpg")
plt.show()
```

```python
### Exceso de velocidad en fecha 1932-01-01
filtro_fecha = df[df["fecha"] == "1932-01-01"]

filtro_fecha.groupby("fecha")["exceso_velocidad"].mean().plot()

plt.title("Exceso promedio en fecha 1932-01-01")

plt.savefig("urban_flow/data/interim/plots/date.jpg")
plt.show()
```
